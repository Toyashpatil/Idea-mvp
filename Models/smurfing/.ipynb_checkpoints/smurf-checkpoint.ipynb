```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
```

```python

file_path = 'smurfDS.xlsx'
df = pd.read_excel(file_path)
```

```python
# datetime to timstamp and  transaction encoding to numberi and AMYt scaler
df['Date_Time'] = pd.to_datetime(df['Date_Time']).astype('int64') // 10**9

label_encoder = LabelEncoder()
df['Transaction_Type'] = label_encoder.fit_transform(df['Transaction_Type'])


scaler = MinMaxScaler()
df[['Withdrawal_Amt', 'Deposit_Amt']] = scaler.fit_transform(df[['Withdrawal_Amt', 'Deposit_Amt']])
```

```python
def create_sequences(data, sequence_length):
    sequences = []
    labels = []
    for i in range(len(data) - sequence_length):
        seq = data[i:i + sequence_length]
        label = data[i + sequence_length][1]  # Withdrawal_Amt as target
        sequences.append(seq)
        labels.append(label)
    return np.array(sequences), np.array(labels)
```

```python
# Prepare data for LSTM model
SEQUENCE_LENGTH = 10
features = df[['Transaction_Type', 'Withdrawal_Amt', 'Deposit_Amt', 'Date_Time']].values.astype('float32')
X, y = create_sequences(features, SEQUENCE_LENGTH)

# Reshape for LSTM input: (samples, time steps, features)
X = X.reshape((X.shape[0], X.shape[1], 4))

# Define LSTM model
model = Sequential()
model.add(tf.keras.layers.Input(shape=(X.shape[1], X.shape[2])))  # Define input shape here
model.add(LSTM(64, return_sequences=True))
model.add(Dropout(0.2))
model.add(LSTM(32, return_sequences=False))
model.add(Dropout(0.2))
model.add(Dense(1, activation='sigmoid'))  # Binary classification (1 = smurfing, 0 = not smurfing)
```

```python
# Compile model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python

# Train model
history = model.fit(X, y, epochs=20, batch_size=16, validation_split=0.2)
```

```python
# Save model
# Save the model in the recommended .keras format
model.save('smurfing_detection_model.keras')
```

```python
test_data = pd.DataFrame({
    'Account_No': ['ACC001', 'ACC001', 'ACC001', 'ACC001', 'ACC001', 'ACC002', 'ACC002', 'ACC002'],
    'Date_Time': pd.to_datetime(['2025-03-18 09:00:00', '2025-03-18 09:05:00', '2025-03-18 09:10:00', 
                                 '2025-03-18 09:15:00', '2025-03-18 09:20:00', '2025-03-18 10:00:00',
                                 '2025-03-18 10:10:00', '2025-03-18 10:15:00']),
    'Transaction_Type': ['Deposit', 'Withdrawal', 'Withdrawal', 'Withdrawal', 'Withdrawal', 'Deposit', 
                         'Withdrawal', 'Withdrawal'],
    'Withdrawal_Amt': [0, 20000, 20000, 20000, 20000, 0, 15000, 10000],
    'Deposit_Amt': [250000, 0, 0, 0, 0, 50000, 0, 0]
})
```

```python
# ✅ Clean up data (remove hidden characters or trailing spaces)
test_data['Transaction_Type'] = test_data['Transaction_Type'].str.strip()

# ✅ Convert Date_Time to integer timestamp (seconds)
test_data['Date_Time'] = test_data['Date_Time'].astype('int64') // 10**9

# ✅ Load the saved model (.keras format)
model = tf.keras.models.load_model('smurfing_detection_model.keras')
```

```python
# ✅ Encode transaction type (use the same LabelEncoder as training)
label_encoder = LabelEncoder()
label_encoder.fit(['Deposit', 'Withdrawal'])  # Must match the training data categories
test_data['Transaction_Type'] = label_encoder.transform(test_data['Transaction_Type'])
```

```python
# Fit MinMaxScaler with DataFrame (match training range)
scaler = MinMaxScaler()
scaler.fit(pd.DataFrame([[0, 0], [250000, 250000]], columns=['Withdrawal_Amt', 'Deposit_Amt']))

# Transform the test data
test_data[['Withdrawal_Amt', 'Deposit_Amt']] = scaler.transform(test_data[['Withdrawal_Amt', 'Deposit_Amt']])
```

```python
# ✅ Function to create sequences
SEQUENCE_LENGTH = 5

def create_sequences(data, sequence_length):
    sequences = []
    for i in range(len(data) - sequence_length + 1):
        seq = data[i:i + sequence_length]
        sequences.append(seq)
    return np.array(sequences), None
```

```python
# ✅ Create sequences for testing
X_test, _ = create_sequences(test_data[['Transaction_Type', 'Withdrawal_Amt', 'Deposit_Amt', 'Date_Time']].values, SEQUENCE_LENGTH)

# ✅ Reshape for LSTM input
X_test = X_test.reshape((X_test.shape[0], X_test.shape[1], 4))
```

```python
# ✅ Predict using the model
predictions = model.predict(X_test)
predicted_classes = (predictions > 0.3).astype(int)
```

```python
# ✅ Display results
for i, pred in enumerate(predicted_classes):
    status = "🔴 Smurfing Detected" if pred == 1 else "✅ Normal"
    print(f"Transaction Sequence {i + 1}: {status}")
```
